Problem \#1

A researcher at the Cleveland Clinic is studying intra-host mutations of SARS-CoV-2 (know this virus?). They recruit 589 COVID-19 patients from the greater Cleveland area and collect nasal swab samples at initial diagnosis and again 2 weeks later. Using deep sequencing, the researcher counts the number of new mutations that were present in the viral Spike (S) gene after the 2-week period for each patient and reports their average across the sample.

Describe the following:
1. Population, sample, and unit of observation
2. Parameter and Statistic
3. Is this a simple random sample?
4. Is this and experimental or observational study?

Answers: 

Problem #1 

1. Population, sample, and unit of observation
- All patients with Covid-19 (or only those within the greater Cleveland area)
- 589 Covid-19 patients
- An individual Patient

2. Parameter and Statistic
- Average number of new spike gene mutations occuring over 2 weeks across the population

3. Is this a simple random sample?

- No! It's a convenience sample of patients within a specific hospital network. There may be bias within patients from Cleveland that could impact viral mutation rates (eg Age)
- Potential Non-response bias 

4. Is this and experimental or observational study?

- Observational Study. They're simply taking samples, not applying treatments.
