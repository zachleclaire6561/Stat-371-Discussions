d. Compute the expected number of heads $E[N]$, the variance $Var(N)$, the standard deviation $sd(N)$, and $E[N^2]-(E[N])^2$. Notice anything similar between these quantities?

$$\begin{align*}
E[N] &= P * N =  \\
&= .45^4 * 0 + 4 * .45^3 * 0.55 * 1 + 6 * .45^2 * 0.55^2 * 2 + 4 * 0.45 * 0.55^3 * 3 + 0.55^4 * 4 
\end{align*}$$

$$\begin{align*}
E[N^2] &= P(N = 0) * 0 + P(N = 1) * 1 + P(N = 2) * 4 + P(N = 3) * 9 + P(N = 4) * 16 \\
&= .45^4 * 0 + 4 * .45^3 * 0.55 * 1 + 6 * .45^2 * 0.55^2 * 4 + 4 * 0.45 * 0.55^3 * 9 + 0.55^4 * 16 \\
&= 5.38
\end{align*}$$

$$\begin{align*}
Var[N] &= P * (1-P) N = 0.55 * 0.45 * 4 = 0.99
\end{align*}$$

Notice that $E[N^2]-(E[N])^2 = 0.99 = Var(N)$. This is a nice way to compute variance.

f. What's the probability of the event that the coin comes up heads an even number of times? 

$$P(N even) = P(N = 0) + P(N = 2) + P(N = 4) = 0.45^4 + 6*0.45^2 *0.55^2 + 0.55^4 \approx 0.5000$$

g. What's the probability that the number of heads is 4 given that the first coin flip was heads? What about if it was tails?

$$P(N = 4 | \text{first flip heads}) = \frac{P(N = 4 \text{ and first flip heads})}{P(\text{first flip heads})} = \frac{0.55^4}{0.55} = 0.55^3 = 0.166375$$

where we use the fact that the event where N = 4 requires that the first flip is heads, so $P(N = 4 \text{ and first flip heads}) = P(N = 4)$

2. Birthday Problem


Suppose you have a group of N people. We say there is a shared birthday if at least two people in the group have the same birthday.

a. Explain why the probablity of this event is:

$$P(\text{shared birthday})  = 1-\frac{(365-1)*(365-2)*...*(365-N)}{365^N}$$

Ans: This is just computing the complement of the event of no shared birthday. We can then just compute the number of no shared by birthdays by counting the number of ways N people can have no birthdays and dividing by the total number of ways N people can have any birthdays.
