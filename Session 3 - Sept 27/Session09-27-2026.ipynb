1. Coin Flips

You flip a biased coin 4 times and count the number of heads, calling it $N$. The biased coin has a 45% chance of comes up tails

a. What's the underlying sample space? How many elements does it have?

b. What type of random variable is the number of heads? What are its parameters?

c. Fill in the following table

|  Number of Heads    |    Probability   |
|----------|:-------------|
| 0        |              |
| 1        |              |  
| 2        |              |   
| 3        |              |   
| 4        |              |   

- Compute it by hand first 
- Check your work with dbinom() function in R (Hint: dbinom(k,n,p) gives you the probability for exactly k successes of n trials with probability p)

d. Compute the expected number of heads $E[N]$, the variance $Var(N)$, the standard deviation $sd(N)$, and $E[N^2]-(E[N])^2$. Notice anything similar between these quantities?

e. What's the probability of the event that the coin comes up heads more than once?

g. What's the probability that the number of heads is 4 given that the first coin flip was heads? What about if it was tails? 

2. Birthday Problem

a. Explain why the probablity of this event is:

$$P(\text{shared birthday})  = 1-\frac{(365-1)*(365-2)*...*(365-N)}{365^N}$$

(Hint: Complement Rule)
