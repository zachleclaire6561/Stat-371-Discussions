- Probability theory describes how likely we will see quantities of samples (like averages or proportions) in random samples of a population given an underlying distribution. For example, if we know that crab claw length is approximately normally distributed with $\mu = 5, \sigma^2 = 1$, then the 68-95-99.7 rule tells us that we will see very few (0.03%) crabs with claw length over 3 standard deviations away from $\mu$, or $\mu \pm 3 * \sigma = 2,8$. 

- Statistics is the reverse problem: it studies how we can learn about the underlying distribution of the population given a sample. For example, if we know that crab claw lengths are approximately normally distributed, then we can just try to estimate the mean and standard deviation to describe the distribution of crabs.
